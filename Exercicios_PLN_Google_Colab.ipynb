## Resumo dos conceitos
- **Corpus:** coleção de documentos.
- **Expressões regulares (Regex):** padrões utilizados para localizar, extrair, substituir e limpar textos de forma eficiente.
- **Tokenização:** divisão do texto em unidades (geralmente palavras ou pontuações).
- **Stopwords:** palavras frequentes que podem ser removidas conforme a tarefa.
- **Stemming:** redução aproximada ao radical.
- **Lematização:** busca da forma canônica/dicionarizada da palavra.
- **POS tagging:** atribuição de categorias gramaticais aos tokens.
- **Bag of Words (BoW):** representação simples do texto por meio da contagem de termos.
- **N-gramas:** sequências de `n` tokens consecutivos que ajudam a preservar o contexto local de grupos de palavras.
- **TF-IDF:** ponderação de termos que equilibra a frequência da palavra no documento com sua raridade no corpus.
- **Embeddings:** representações vetoriais contínuas que capturam o significado semântico e padrões de uso das palavras.

**Importante:** não existe uma única sequência de pré-processamento adequada a todas as tarefas de PLN. Cada escolha deve ser justificada pelo objetivo do projeto.
